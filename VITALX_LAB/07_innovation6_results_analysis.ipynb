````python
# ============================================
# ASTROVITAL AI: VITALX CORE V1
# Notebook 07 — INNOVATION 06 6 Part 2
# Biomarker Deep Analysis
# Developer: Gouragopal Mohapatra
# GitHub: GOURGOPAL618
# © 2026 Gouragopal Mohapatra — All Rights Reserved
# ============================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
import joblib
import warnings
warnings.filterwarnings('ignore')

pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.4f}'.format)
plt.style.use('dark_background')
sns.set_palette('coolwarm')
np.random.seed(42)

# Load Dataset
df_biomarkers = pd.read_csv(
    '../DATA_VAULT/MISSION_READY_DATA/astrovital_biomarkers_v1.csv'
)

df_clean = pd.read_csv(
    '../DATA_VAULT/MISSION_READY_DATA/astrovital_signal_clean_v1.csv'
)

# Load saved Model
rf_model = joblib.load('../MODEL_HANGAR/innovation6_biomarker_model.pkl')
le = joblib.load('../MODEL_HANGAR/innovation6_label_encoder.pkl')
final_biomarkers = joblib.load('../MODEL_HANGAR/innovation6_biomarker_list.pkl')

print("=" * 57)
print("  ASTROVITAL AI : VITALX CORE V1")
print("  NOTEBOOK 07 - INNOVATION 6 Result Analysis")
print("  Because Every Heartbeat In Space Matters.")
print("=" * 57)
print(f"✅ Biomarker Dataset:  {df_biomarkers.shape}")
print(f"✅ Clean Dataset:      {df_clean.shape}")
print(f"✅ Models Loaded:      RF + Encoder + Biomarker list")
print(f"✅ Final Biomarkers:   {len(final_biomarkers)}")
print(f"\n📊 Dataset Preview:")
print(df_biomarkers.head())
````

## RESULT ANALYSIS PLAN

### WHAT THIS NOTEBOOK COVERS:
| Section | Analysis | Purpose |
|---|---|---|
| **Statistical** | Per-class biomarker stats | Understand health patterns |
| **Visual** | Biomarker distributions | Paper-ready figures |
| **Comparative** | GREEN vs YELLOW vs RED | Clinical differentiation |
| **Correlation** | Biomarker relationships | Redundancy analysis |

### KEY RESEARCH QUESTIONS:
> 1. How do top biomarkers differ across health classes?
> 2. Which biomarkers best separate GREEN from RED?
> 3. Are any biomarkers redundant?
> 4. What clinical thresholds emerge from data?

````python
# ============================================
# Per Class Biomarker Statistics
# ============================================

print("=" * 57)
print("  Biomarker Statistics - Per Health Class")
print("=" * 57)

for biomarker in final_biomarkers[:5]:
    print(f"\n📊 {biomarker}:")
    print(f"{'Class':<10} {'Mean':>10} {'Std':>10} {'Min':>10} {'Max':>10}")
    print("-" * 48)
    for cls in ['GREEN', 'YELLOW', 'RED']:
        subset = df_clean[df_clean['Health_Status'] == cls][biomarker]
        if len(subset) > 0:
            print(f"{cls:<10} {subset.mean():>10.4f} "
                  f"{subset.std():>10.4f} "
                  f"{subset.min():>10.4f} "
                  f"{subset.max():>10.4f}")
````

````python
# ============================================
# Biomarker Distribution — Per Health Class
# ============================================

top_4 = final_biomarkers[:4]
fig, axes = plt.subplots(2, 2, figsize = (14, 10))
fig.suptitle(
    'ASTROVITAL AI : VITALX CORE V1\n'
    'INNNOVATION 6 - TOP BIOMARKER DISTRIBUTION BY HEALTH CLASS',
    color = 'gold', fontsize = 14, fontweight = 'bold'
)

class_colors = {
    'GREEN': '#00E676',
    'YELLOW': '#FFD700',
    'RED': '#D30000'
}

for idx, biomarker in enumerate(top_4):
    row, col = divmod(idx, 2)
    for cls, color in class_colors.items():
        subset = df_clean[
            df_clean['Health_Status'] == cls
        ][biomarker]
        if len(subset) > 0:
            axes[row, col].hist(
                subset, bins = 25,
                alpha = 0.6, color = color,
                label = cls, edgecolor = 'none'
            )
    
    axes[row, col].set_title(biomarker,
                             color = 'white', fontsize = 10)
    axes[row, col].tick_params(color = 'white')
    axes[row, col].legend(fontsize = 8)
    axes[row, col].set_ylabel('Count', color = 'white')

plt.tight_layout()
plt.show()
print("✅ Biomarker distribution visualization complete!")
````

````python
# ============================================
# Boxplot — Health Class Comparison
# ============================================

top_6 = final_biomarkers[:6]

fig, axes = plt.subplots(2, 3, figsize = (16, 10))
fig.suptitle(
    'ASTROVITAL AI : VITALX CORE V1\n'
    'INNOVATION 6 - BIOMARKER VS HEALTH STATUS',
    color = 'gold', fontsize = 14, fontweight = 'bold'
)

for idx, biomarker in enumerate(top_6):
    row, col = divmod(idx, 3)

    data_by_class = [
        df_clean[df_clean['Health_Status'] == cls][biomarker].values
        for cls in ['GREEN', 'YELLOW', 'RED']
    ]

    bp = axes[row, col].boxplot(
        data_by_class,
        patch_artist = True,
        labels = ['GREEN', 'YELLOW', 'RED']
    )

    colors_box = ['#00E676', '#FFD300', '#D30000']
    for patch, color in zip(bp['boxes'], colors_box):
        patch.set_facecolor(color)
        patch.set_alpha(0.7)

    axes[row, col].set_title(biomarker, color = 'white', fontsize = 10)
    axes[row, col].tick_params(colors = 'white', labelsize = 8)

plt.tight_layout()
plt.show()
print("✅ Boxplot comparison complete!")
````

## PATTERN ANALYSIS - WHAT THE DATA REVEALS

### KEY OBSERVATIONS:
**GREEN Class (Safe 🟢):**
> - Heart Rate within normal range
> - Adequate sleep hours maintained
> - Oxygen Saturation stable
> - Low Overall risk score

**YELLOW Class (Warnings ⚠️):**
> - 1-2 biomarkers showing deviation
> - Early signs of physiological stress
> - Interventation recommended soon

**RED Class (Critical ❗):**
> - Multiple biomarkers simultaneously abnormal
> - Significant physiological deterioration
> - Immediate medical interventation required

### CLINICAL THRESHOLD INSIGHTS:
> The boxplots reveal natural separation points
> between health classes - these thresholds will
> directly inform **INNOVATION 1** Decision Tree rules.

````python
# ============================================ 
# 3D Biomarker Space — Top Feature Visualization
# ============================================

from mpl_toolkits.mplot3d import Axes3D
import matplotlib.pyplot as plt

top_3 = final_biomarkers[:3]
plot_cols = top_3 + ['Health_Status']
df_plot = df_clean[plot_cols].copy()

# Sample for clarity
df_sample = df_plot.sample(min(300, len(df_plot)), random_state=42)

# Color mapping
palette = {
    'GREEN': '#00E676',
    'YELLOW': '#FFD700',
    'RED': '#FF1744'
}

# Create figure with dark "space" theme
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

fig.patch.set_facecolor('black')
ax.set_facecolor('black')

# Plot each class
for status, color in palette.items():
    subset = df_sample[df_sample['Health_Status'] == status]
    
    ax.scatter(
        subset[top_3[0]],
        subset[top_3[1]],
        subset[top_3[2]],
        c=color,
        label=status,
        s=40,
        alpha=0.8
    )

# Labels
ax.set_xlabel(top_3[0], color='white')
ax.set_ylabel(top_3[1], color='white')
ax.set_zlabel(top_3[2], color='white')

# Title
plt.title(
    'ASTROVITAL AI : VITALX CORE V1\n3D Biomarker Space',
    color='gold',
    fontsize=12,
    fontweight='bold'
)

# Style tweaks
ax.tick_params(colors='white')
ax.legend(facecolor='black', edgecolor='white')

plt.tight_layout()
plt.show()

print("🚀 Space-style 3D biomarker visualization ready!")
print("✅ Biomarker separation visible across classes!")
````

````python
# ============================================
# Clinical Threshold Analysis
# ============================================

print("=" * 57)
print("  Clinical Thresholds - Derived from Data")
print("=" * 57)

thresholds_derived = {}

for biomarker in final_biomarkers:
    green = df_clean[
        df_clean['Health_Status'] == 'GREEN'
    ][biomarker]
    red = df_clean[
         df_clean['Health_Status'] == 'RED'
    ][biomarker]

    if len(green) > 0 and len(red) > 0:
         green_upper = green.quantile(0.95)
         red_lower = red.quantile(0.05)
         threshold = (green_upper + red_lower) / 2 

         thresholds_derived[biomarker] = round(threshold, 4)
         
         print(f"\n📊 {biomarker}:")
         print(f"   GREEN Upper (95th):   {green_upper:.4f}")
         print(f"   RED Lower (5th):      {red_lower:.4f}")
         print(f"   ➡️ Derived Threshold: {threshold:.4f}")
         
print(f"\n✅ Clinical Thresholds Derived!")
print(f"✅ These will inform INNOVATION 1 DECISION TREE!")
````

````python
# ============================================
# Innovation 6 — Complete Summary Graph (FIXED)
# ============================================

fig, axes = plt.subplots(2, 2, figsize = (16, 12))
fig.suptitle(
    'ASTROVITAL AI : VITALX CORE V1\n'
    'INNOVATION 6 - COMPLETE BIOMARKER ANALYSIS SUMMARY',
    color='gold', fontsize=14, fontweight='bold'
)

# ✅ Use SAME features as training
feature_names = [
    'Heart_rate_bpm', 'Systolic_BP_mmHg',
    'Diastolic_BP_mmHg', 'Sleep_Hours',
    'Oxygen_Saturation', 'Muscle_Loss_Percent',
    'Bone_Density_Loss', 'Radiation_mSv',
    'Immune_Score'
]

# 1. Importance Ranking
importance_scores = rf_model.feature_importances_

imp_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': importance_scores
}).sort_values('Importance', ascending=True)

colors_imp = ['red' if v > 0.10 else
              'yellow' if v > 0.05 else 'lime'
              for v in imp_df['Importance']]

axes[0, 0].barh(
    imp_df['Feature'],
    imp_df['Importance'],
    color=colors_imp, alpha=0.8, edgecolor='white'
)
axes[0, 0].set_title('BIOMARKER IMPORTANCE RANKING', fontsize=10, color='white')
axes[0, 0].tick_params(color='white', labelsize=7)

# 2. Health Class Distribution
class_counts = df_clean['Health_Status'].value_counts()
axes[0, 1].pie(
    class_counts.values,
    labels=class_counts.index,
    colors=['#00E676', '#FFD700', '#D30000'],
    autopct='%1.1f%%',
    startangle=90
)
axes[0, 1].set_title('HEALTH CLASS DISTRIBUTION', color='white', fontsize=10)

# 3. Top biomarker per class mean
top_3 = feature_names[:3]   # safe selection
class_means = df_clean.groupby('Health_Status')[top_3].mean()

x = np.arange(len(top_3))
width = 0.25

for i, (cls, color) in enumerate(zip(
    ['GREEN', 'YELLOW', 'RED'],
    ['#00E676', '#FFD700', '#D30000']
)):
    if cls in class_means.index:
        axes[1,0].bar(
            x + i*width,
            class_means.loc[cls],
            width, label=cls,
            color=color, alpha=0.8,
            edgecolor='white'
        )

axes[1,0].set_title('TOP 3 BIOMARKER BY CLASS', color='white', fontsize=10)
axes[1,0].set_xticks(x + width)
axes[1,0].set_xticklabels(
    [b.replace('_bpm','').replace('_mmHg','') for b in top_3],
    color='white', fontsize=8
)
axes[1,0].tick_params(colors='white')
axes[1,0].legend(fontsize=8)

# 4. Derived thresholds
if thresholds_derived:
    thresh_names = list(thresholds_derived.keys())[:6]
    thresh_values = [thresholds_derived[k] for k in thresh_names]

    axes[1, 1].barh(
        thresh_names, thresh_values,
        color='cyan', alpha=0.8,
        edgecolor='white'
    )
    axes[1, 1].set_title('DERIVED CLINICAL THRESHOLDS', color='white', fontsize=10)
    axes[1, 1].tick_params(colors='white', labelsize=8)

plt.tight_layout()
plt.show()

print("✅ Complete INNOVATION 6 summary visualization!")
````
