````python
# ============================================
# MASTER PERFORMANCE CHART
# ============================================

fig, axes = plt.subplots(2, 3, figsize=(16, 10))
fig.patch.set_facecolor('#0B0F1A')
fig.suptitle(
    'ASTROVITAL AI — VITALX CORE V1\n'
    'Complete Project Results — All 3 Innovations',
    color='gold', fontsize=14, fontweight='bold'
)

# Panel 1 — Innovation 3 R2
inn3_params = ['HR', 'SBP', 'DBP', 'Sleep',
               'SpO2', 'Muscle', 'Bone',
               'Radiation', 'Immune']

inn3_r2 = [0.82, 0.84, 0.83, 0.81,
           0.85, 0.80, 0.82, 0.84, 0.83]

axes[0, 0].bar(inn3_params, inn3_r2,
               color = '#00E676', alpha = 0.85,
               edgecolor = 'white')

axes[0, 0].axhline(y = 0.80, color = 'gold', linestyle = '--', linewidth = 1.5,
                   label = 'Target R2 = 0.80')

axes[0, 0].set_title('INNOVATION 03 - R2 per Sensor',
                     color = '#00E676', fontsize = 10,
                     fontweight = 'bold')

axes[0, 0].tick_params(colors = 'white', labelsize = 7)
axes[0, 0].set_ylim([0.75, 0.90])
axes[0, 0].legend(fontsize = 8)
axes[0, 0].set_facecolor('#0B0F1A')

# Panel 2 — Innovation 6 feature importance
top_features = features[:6] if len(features) >= 6 \
                else features
imp = cdss_model.feature_importances_[:len(top_features)]
imp_df = pd.DataFrame({
    'Feature': top_features,
    'Importance': imp
}).sort_values('Importance', ascending = False)

short_names = [f.replace('_bpm', '').replace('_mSv','')
               .replace('_Loss', '').replace('_Score','')
               .replace('_Risk', '').replace('_Percent','')
               for f in imp_df['Feature']]

colors_imp = ['#D30000' if v > 0.15 else
              '#FFD700' if v > 0.08 else '#B388FF'
              for v in imp_df['Importance']]

axes[0, 1].barh(short_names, imp_df['Importance'],
               color=colors_imp, alpha=0.85,
               edgecolor='white')

axes[0, 1].set_title('Innovation 6 — Feature Importance',
                     color='#B388FF', fontsize=10,
                     fontweight='bold')

axes[0, 1].tick_params(colors = 'white', labelsize = 7)
axes[0, 1].set_facecolor('#0B0F1A')

# Panel 3 — Innovation 1 per class
classes = le_cdss.classes_
precision_vals = [0.96, 0.56, 0.84]
recall_vals    = [0.91, 0.76, 0.81]
f1_vals        = [0.94, 0.65, 0.82]

x = np.arange(len(classes))
w = 0.25

axes[0, 2].bar(x-w, precision_vals, w,
              label='Precision',
              color='#00C8FF', alpha=0.8,
              edgecolor='white')

axes[0, 2].bar(x,   recall_vals,    w,
              label='Recall',
              color='#FFD700', alpha=0.8,
              edgecolor='white')

axes[0, 2].bar(x+w, f1_vals,        w,
              label='F1',
              color='#00E676', alpha=0.8,
              edgecolor='white')

axes[0, 2].set_title('Innovation 1 — Per Class Metrics',
                     color='#00C8FF', fontsize=10,
                     fontweight='bold')

axes[0, 2].set_xticks(x)
axes[0, 2].set_xticklabels(classes, color = 'white')
axes[0, 2].tick_params(colors = 'white')
axes[0, 2].legend(fontsize = 8)
axes[0, 2].set_facecolor('#0B0F1A')

# Panel 4 — RED recall improvement
baseline = [10.34, 18.00, 12.50]
production = [75.86, 64.71, 56.41]
metrics_r = ['RED\nRecall', 'RED\nF1', 'RED\nPrecision']

x2 = np.arange(len(metrics_r))

axes[1,0].bar(x2-0.2, baseline, 0.35,
              label='Baseline', color='#D30000',
              alpha=0.7, edgecolor='white')

axes[1,0].bar(x2+0.2, production, 0.35,
              label='Production', color='#00E676',
              alpha=0.85, edgecolor='white')

axes[1,0].set_title('RED Class — Baseline vs Production',
                     color='white', fontsize=10,
                     fontweight='bold')

axes[1, 0].set_xticks(x2)
axes[1, 0].set_xticklabels(metrics_r, color = 'white')
axes[1, 0].tick_params(colors = 'white')
axes[1, 0].legend(fontsize = 8)
axes[1, 0].set_facecolor('#0B0F1A')

# Panel 5 — Health distribution

sizes = [479, 403, 118]
colors = ['#00E676', '#FFD700', '#FF4444']
lbls   = ['GREEN\n47.9%', 'YELLOW\n40.3%',
          'RED\n11.8%']

wedges, texts = axes[1, 1].pie(
    sizes, colors = colors, startangle = 90,
    wedgeprops = dict(width = 0.5,
                      edgecolor = '#0B0F1A',
                      linewidth = 2))

axes[1, 1].text(0, 0, 'V2\nDataset',
               ha='center', va='center',
               color='white', fontsize=9,
               fontweight='bold')

for text, lbl, col in zip(texts, lbls, colors):
    text.set_text(lbl)
    text.set_color(col)
    text.set_fontsize(8)

axes[1, 1].set_title('Dataset Health Distribution',
                     color='white', fontsize=10,
                     fontweight='bold')

axes[1, 1].set_facecolor('#0B0F1A')

# Panel 6 — Overall summary
axes[1, 2].set_facecolor('#0B0F1A')
axes[1, 2].axis('off')
axes[1, 2].set_title('PROJECT SUMMARY',
                     color='white', fontsize=10,
                     fontweight='bold')
summary = [
    ('Notebooks',    '11 Complete',  '#00C8FF'),
    ('Innovations',  '3 Working',    '#00E676'),
    ('Models',       '6 Saved',      '#00E676'),
    ('Accuracy',     '85.25%',       '#00E676'),
    ('RED Recall',   '75.86%',       '#FFD700'),
    ('CDSS Gap',     '12yr Bridged', '#FFD700'),
    ('Edge Ready',   'No Cloud',     '#00E676'),
    ('Status',       'COMPLETE ✅',  '#00E676'),
]

for i, (k, v, col) in enumerate(summary):
    yp = 0.92 - i*0.12

    axes[1, 2].text(0.05, yp, f'{k}:',
                   transform=axes[1,2].transAxes,
                   color='white', fontsize=8,
                   va='center') 
    
    axes[1, 2].text(0.55, yp, v,
                   transform=axes[1,2].transAxes,
                   color=col, fontsize=8,
                   fontweight='bold', va='center')

plt.tight_layout()
plt.show()
print("✅ Master Performance Chart Complete!")
````
